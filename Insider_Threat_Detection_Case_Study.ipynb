```python
# Load & Prepare Threat Labels in insiders.csv
import pandas as pd

# 1. Load ground truth
answers_csv_path = "./dataset/answers/insiders.csv"
df_insiders = pd.read_csv(answers_csv_path)

# 2. Filter for version 4.1
# CERT stores dataset versions as floats or strings (e.g., 4.1 or '4.1')
df_insiders['dataset_str'] = df_insiders['dataset'].astype(str)
threats = df_insiders[df_insiders['dataset_str'].str.startswith('4.1')].copy()

# 3. Convert start and end timestamps into date objects
threats['start_dt'] = pd.to_datetime(threats['start']).dt.date
threats['end_dt'] = pd.to_datetime(threats['end']).dt.date

print(f"Threat scenarios detected for release 4.1: {len(threats)}\n")
print(threats[['dataset', 'scenario', 'user', 'start_dt', 'end_dt']])
```

```python
# Phase 3 Feature Engineering and Log Aggregation
import pandas as pd
import numpy as np

# 1. Process Logon Activity
logon_path = "./dataset/r4.1/logon.csv"  # verify exact extracted path
print("Loading logon logs...")

# If the file is huge, read in chunks or specify columns to save RAM
cols = ['id', 'date', 'user', 'pc', 'activity']
df_logon = pd.read_csv(logon_path, usecols=cols)

# Convert timestamp
df_logon['timestamp'] = pd.to_datetime(df_logon['date'], format='%m/%d/%Y %H:%M:%S')
df_logon['day'] = df_logon['timestamp'].dt.date
df_logon['hour'] = df_logon['timestamp'].dt.hour
df_logon['is_weekend'] = df_logon['timestamp'].dt.weekday >= 5
df_logon['is_after_hours'] = (df_logon['hour'] < 7) | (df_logon['hour'] > 19)

# Aggregate daily logon stats per user
logon_feats = df_logon.groupby(['user', 'day']).agg(
    total_logons=('activity', lambda x: (x == 'Logon').sum()),
    total_logoffs=('activity', lambda x: (x == 'Logoff').sum()),
    distinct_pcs=('pc', 'nunique'),
    after_hours_logons=('is_after_hours', 'sum'),
    weekend_logons=('is_weekend', 'sum')
).reset_index()

# 2. Process Device (USB) Activity
device_path = "./dataset/r4.1/device.csv"
print("Loading device logs...")
df_device = pd.read_csv(device_path)
df_device['timestamp'] = pd.to_datetime(df_device['date'], format='%m/%d/%Y %H:%M:%S')
df_device['day'] = df_device['timestamp'].dt.date
df_device['hour'] = df_device['timestamp'].dt.hour
df_device['is_after_hours'] = (df_device['hour'] < 7) | (df_device['hour'] > 19)

device_feats = df_device.groupby(['user', 'day']).agg(
    usb_connects=('activity', lambda x: (x == 'Connect').sum()),
    usb_disconnects=('activity', lambda x: (x == 'Disconnect').sum()),
    after_hours_usb=('is_after_hours', 'sum')
).reset_index()

# 3. Merge User-Day Features
features_df = pd.merge(logon_feats, device_feats, on=['user', 'day'], how='left').fillna(0)
print("Aggregated dataset shape:", features_df.shape)
print(features_df.head())
```

```python
import pandas as pd

# 1. Ensure 'day' column is in Python date format for exact matching
features_df['day'] = pd.to_datetime(features_df['day']).dt.date

# 2. Initialize all user-day rows as normal/benign (0)
features_df['label'] = 0

# 3. Mark rows falling within documented threat intervals as 1
for _, threat in threats.iterrows():
    start = threat['start_dt'] if 'start_dt' in threat else threat['start_date']
    end = threat['end_dt'] if 'end_dt' in threat else threat['end_date']
    
    mask = (
        (features_df['user'] == threat['user']) &
        (features_df['day'] >= start) &
        (features_df['day'] <= end)
    )
    features_df.loc[mask, 'label'] = 1

# 4. Display label distribution
print("Phase 4 Complete: Ground-truth labels successfully defined.")
print("\n--- Label Distribution ---")
print(features_df['label'].value_counts())
print(f"\nThreat ratio: {(features_df['label'].mean() * 100):.4f}%")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

# 1. Feature selection (exclude identifiers and target column)
feature_cols = [col for col in features_df.columns if col not in ['user', 'day', 'label']]
print(f"Features used for training ({len(feature_cols)}):", feature_cols)

X = features_df[feature_cols].fillna(0)
y = features_df['label']

# 2. Stratified train-test split (preserves threat proportion in both sets)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

# 3. Standardize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Train Logistic Regression with balanced class weights
clf = LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42)
clf.fit(X_train_scaled, y_train)

# 5. Predictions & Evaluation
y_pred = clf.predict(X_test_scaled)
y_prob = clf.predict_proba(X_test_scaled)[:, 1]

print("\n" + "=" * 45)
print("CONFUSION MATRIX:")
print(confusion_matrix(y_test, y_pred))

print("\n" + "=" * 45)
print("CLASSIFICATION REPORT:")
print(classification_report(y_test, y_pred, digits=4))

print("=" * 45)
print(f"ROC-AUC SCORE: {roc_auc_score(y_test, y_prob):.4f}")

# 6. Model Interpretability (Feature Coefficients)
coef_df = pd.DataFrame({
    'Feature': feature_cols,
    'Coefficient': clf.coef_[0]
}).sort_values(by='Coefficient', ascending=False)

print("\n" + "=" * 45)
print("FEATURE COEFFICIENTS (ODDS RATIO IMPACT):")
print(coef_df.to_string(index=False))
```
